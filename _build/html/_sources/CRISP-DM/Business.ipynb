# Business Understanding

## 1. Mengidentifikasi Masalah

Mengumpulkan **200 artikel berita Indonesia** dari [Detik.com](https://www.detik.com/) sebagai dataset untuk tugas mata kuliah **Pencarian dan Penambangan Web (PPW)**. Proses pengumpulan data disusun mengikuti metodologi **CRISP-DM**.

Fokus permasalahan adalah melakukan **kategorisasi berita Indonesia** ke dalam dua kategori, yaitu **sport** dan **finance**. Tahapan pengolahan dan klasifikasi data akan dilanjutkan pada tahap-tahap CRISP-DM berikutnya.

## 2. Menentukan Tujuan

Tujuan dari tahap ini adalah:

* Membangun dataset berlabel yang bersih dan terdiri dari **200 artikel berita**.
* Dataset terdiri dari **100 artikel kategori sport** dan **100 artikel kategori finance**.
* Dataset memiliki struktur kolom:

  * `id`
  * `isi_berita`
  * `label`
  * `url`
* Menyiapkan data agar dapat digunakan pada tahap **Data Preparation, Modeling, Evaluation,** dan **Deployment** pada tugas selanjutnya.

## 3. Mengidentifikasi Pihak Yang Berkepentingan

Pihak-pihak yang berkepentingan dalam proses penelitian ini meliputi:

* **Mahasiswa**: menyusun dan mengolah dataset sebagai bagian dari tugas mata kuliah PPW.
* **Dosen pengampu**: mengevaluasi proses dan hasil pengerjaan tugas.
* **Pembaca Jupyter Book / GitHub Pages**: melihat dokumentasi proses pengumpulan dan pengolahan data.

## 4. Menentukan Kriteria Keberhasilan

Keberhasilan proses pengumpulan data ditentukan berdasarkan beberapa kriteria berikut:

* Dataset memiliki tepat **200 baris data**.
* Terdapat **100 artikel dengan label `sport`** dengan ID 1–100.
* Terdapat **100 artikel dengan label `finance`** dengan ID 101–200.
* Tidak terdapat nilai kosong pada kolom **`isi_berita`**.
* Nilai `id` tersusun secara berurutan dari **1 sampai 200**.
* Dataset dapat dimuat tanpa error.
* Dataset memiliki struktur kolom yang sesuai, yaitu:
  `id`, `isi_berita`, `label`, dan `url`.

## 5. Menilai Situasi, Batasan, dan Etika

Proses pengumpulan data dilakukan dengan memperhatikan aspek teknis, etika, dan kesopanan dalam melakukan crawling.

* **Politeness**: setiap request diberikan jeda minimal **1 detik** untuk mengurangi beban terhadap server.
* **User-Agent**: request menggunakan header **User-Agent** yang menyerupai browser umum.
* **Tujuan penggunaan**: data yang dikumpulkan hanya digunakan untuk **keperluan pembelajaran dan tugas mata kuliah PPW**.
* **Sumber data**: URL artikel diperoleh secara otomatis melalui halaman indeks kategori resmi Detik.com, yaitu kategori **sport** dan **finance**, bukan dikumpulkan atau ditulis secara manual.
* **Batasan data**: jumlah data dibatasi sebanyak **200 artikel** sesuai kebutuhan tugas.
